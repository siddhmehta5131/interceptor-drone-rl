# 03 - Flight-controller training (P7.3)

PPO with the asymmetric actor-critic: actor 33 observations, critic 49 (16 privileged physics/wind values). Domain randomisation and Dryden wind are on. Three seeds, resumable runs.

```python
import sys, os
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))
print('repo root:', ROOT)
```

```python
from training.algos import load_training_config
from training.runner import train, load_run_model
cfg = load_training_config()
BUDGET = int(cfg['budgets']['fc_train_steps'])
BUDGET
```

```python
for seed in (0, 1, 2):
    run_id = f'p7_fc_asym_s{seed}'
    out = train(run_id, {
        'task': 'fc', 'algo': 'ppo', 'seed': seed,
        'total_steps': BUDGET, 'checkpoint_every': 20000,
        'asymmetric': True, 'actor_obs_dim': 33,
        'fc': {'domain_randomization': True, 'wind': True}})
    print(run_id, out['num_timesteps'])
```

```python
model = load_run_model(ROOT / 'runs' / 'p7_fc_asym_s0')
print(type(model.policy).__name__, model.policy.actor_obs_dim, model.num_timesteps)
```
